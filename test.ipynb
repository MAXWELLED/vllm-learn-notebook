```python
from functools import wraps

def singleton(cls):
    instances = {}
    @wraps(cls)
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@singleton
class DatabaseConnection:
    pass  # 全局唯一实例
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, Ellipse

fig, ax = plt.subplots()

# 左下角(1,1)，宽3，高2
rect = Rectangle(
    xy=(1, 1),
    width=3,
    height=2,
    linewidth=2,
    edgecolor="red",    # 边框颜色
    facecolor="none",    # 填充透明
    angle=30,            # 旋转角度
    rotation_point="xy"
)
circle = Circle(
    xy=(3, 2),          # 圆心
    radius=0.5,         # 半径
    linewidth=2,
    edgecolor="blue",   # 边框颜色
    facecolor="none"    # 填充透明
)
ell = Ellipse(
    xy=(4, 3),          # 圆心
    width=1,            # 宽度
    height=0.5,         # 高度
    linewidth=2,
    edgecolor="yellow",  # 边框颜色
    facecolor="none"    # 填充透明
)
x_labels = [1, 2, 3, 4, 5]
y_labels = [[1, 3, 5], [3, 1, 4], [1, 5, None], [4, 4, 4], [0, 2, 5]]
ax.set_prop_cycle(color=["red", "blue", "green", "orange", "purple"])  # 设置颜色循环
ax.plot([x for x in x_labels], [y for y in y_labels], linewidth=1)  # 绘制一条直线
ax.add_patch(rect)  # 必须add_patch添加到坐标轴
ax.add_patch(circle)
ax.add_patch(ell)
ax.set_xlim(0, 6)
ax.set_ylim(0, 5)
plt.show()
```
